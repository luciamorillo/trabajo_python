```python
import geopandas as gpd
import matplotlib.pyplot as plt

# 1. Cargamos el mapa moderno directamente con la URL oficial
world = gpd.read_file("https://naturalearth.s3.amazonaws.com/110m_cultural/ne_110m_admin_0_countries.zip")

# 2. Filtramos para quedarnos solo con Europa
europa = world[world['CONTINENT'] == 'Europe']

# 3. Dibujamos el mapa
europa.plot(color='lightblue', edgecolor='black', figsize=(8, 6))

# 4. Mostramos el resultado
plt.title("Mi primer mapa oficial con Geopandas y Positron")
plt.show()
```
